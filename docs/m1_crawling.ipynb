```python
import json
import pandas as pd
import matplotlib.pyplot as plt

with open("../reports/m1_benchmark.json", "r") as f:
    stats = json.load(f)

df_stats = pd.DataFrame([stats]).T.rename(columns={0: "Giá trị"})
display(df_stats)

# Biểu đồ so sánh dung lượng trước và sau khi lọc ảnh
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(["Trước khi xóa ảnh", "Sau khi xóa ảnh"], 
       [stats["avg_size_before_kb"], stats["avg_size_after_kb"]], 
       color=["#ef5350", "#42a5f5"])
ax.set_ylabel("Dung lượng trung bình (KB)")
ax.set_title("Hiệu quả giảm dung lượng lưu trữ khi lọc file ảnh (M1)")
plt.tight_layout()
plt.show()
```
