```python
import json
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu thống kê từ file JSON
with open("../reports/m2_benchmark.json", "r", encoding="utf-8") as f:
    m2_stats = json.load(f)

# Hiển thị bảng tổng hợp các chỉ số cốt lõi của Milestone 2
df_m2 = pd.DataFrame([m2_stats]).T.rename(columns={0: "Giá trị (Mock)"})
display(df_m2)

# Biểu đồ mẫu so sánh số lượng Elements và BibTeX entries trước và sau khi Deduplication
categories = ["LaTeX Elements", "BibTeX Entries"]
before_dedup = [m2_stats["elements_before_dedup"], m2_stats["bibtex_before_dedup"]]
after_dedup = [m2_stats["elements_after_dedup"], m2_stats["bibtex_after_dedup"]]

x = range(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar([i - width/2 for i in x], before_dedup, width, label="Trước Dedup", color="#ffa726")
ax.bar([i + width/2 for i in x], after_dedup, width, label="Sau Dedup", color="#66bb6a")

ax.set_ylabel("Số lượng bản ghi")
ax.set_title("Hiệu quả khử trùng lặp (Deduplication) ở Milestone 2 (Mock)")
ax.set_xticks(list(x))
ax.set_xticklabels(categories)
ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
